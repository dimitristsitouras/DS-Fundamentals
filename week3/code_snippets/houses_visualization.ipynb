```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('houses.csv')
df.columns = df.columns.str.strip()

sns.boxplot(data=df, x='number of rooms', y='selling price', color='red')
plt.show()
```

I compared the number of rooms with the selling price in the houses dataset, using a boxplot. I used a boxplot because it is ideal for discrete values like the number of rooms.

We observe a clear positive trend where the median selling price rises as the number of rooms increases from 5 to 8, with the median price ranging from $30,500 to $41,500. Moreover, houses with 9 and 10 rooms separate from the rest, reaching above $80,000.
This pattern shows us that the number of rooms plays a primary role in determining the value of the house. This helps buyers and agents see if a house has a good price.
